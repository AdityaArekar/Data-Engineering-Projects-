```python
from pyspark import SparkContext, SparkConf
import collections
```

```python
conf = SparkConf().setMaster('local').setAppName('WeatherAnalysis')
sc = SparkContext(conf = conf)
```

```python
def parseLines(line):
    fields = line.split(',')
    stationID = fields[0]
    entryType = fields[2]
    temperature = float(fields[3]) * 0.1 * (9.0/5.0) + 32.0
    return (stationID, entryType, temperature)
```

```python
lines = sc.textFile("C:/Users/aaa17/Documents/GitHub/Data-Engineering-Projects-/Udemy Projects/spark/weather_analysis/1800.csv")
parsedLines = lines.map(parseLines)
```

```python
print(parsedLines.collect())
```

```python
minTemps = parsedLines.filter(lambda x: 'TMIN' in x[1])
print(minTemps.collect())
```

```python
stationTemps = parsedLines.map(lambda x: (x[0], x[2]))
print(stationTemps.collect())
```

```python
minTemps = stationTemps.reduceByKey(lambda x, y: max(x,y))
results = minTemps.collect()
```

```python
for result in results:
    print (result[0] + "\t{:.2f}F".format(result[1]) )
```
